```python
from sklearn.feature_extraction.text import CountVectorizer
sentence ="data science and ai genai has great career ahead data data"
vectorizer=CountVectorizer()
vector=vectorizer.fit_transform([sentence])
vector.toarray()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
sentence="data data science and ai genai has great career ahead"
vectorizer=TfidfVectorizer()
vector=vectorizer.fit_transform([sentence])
vector.toarray()
```

```python
import nltk
from nltk import pos_tag,word_tokenize,RegexpParser



text="full stack datascience,generative ai ,agentic ai,llm model keep increase by different company"

tokens=word_tokenize(text)
tagged_tokens=pos_tag(tokens)

chunk_grammar=r"""
 NP:{<DT?><JJ>*<NN>} #Noun Phrase
 VP:{<VB.*><NP|PP>*} #Verb Phrase
 PP:{<IN><NP>} #Prepositional Phrase
"""

chunk_parser=RegexpParser(chunk_grammar)
chunked=chunk_parser.parse(tagged_tokens)
print(chunked)
chunked.draw()
```
