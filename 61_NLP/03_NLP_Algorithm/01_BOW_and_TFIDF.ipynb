```python
paragraph="As digital systems become faster and more connected, a quiet but powerful shift is taking place in how data is processed: edge computing. Traditionally, most data generated by devices such as smartphones, sensors, and cameras is sent to centralized cloud servers for processing. While effective, this approach can introduce latency, consume bandwidth, and raise privacy concerns. Edge computing changes this model by moving computation closer to where the data is generated—at the “edge” of the network.This shift is especially important for applications that require real-time responses. Self-driving cars, smart traffic systems, and industrial robots cannot afford delays caused by sending data back and forth to distant data centers. By processing data locally, edge computing enables faster decision-making, improves reliability, and reduces dependence on continuous internet connectivity. It also helps in handling the massive data volumes produced by Internet of Things (IoT) devices, preventing networks from becoming overloaded.Another key advantage of edge computing is improved data privacy and security. Sensitive information, such as video feeds or health data, can be analyzed locally without being transmitted to the cloud, reducing the risk of large-scale data breaches. As AI models become smaller and more efficient, they can be deployed directly on edge devices, combining intelligence with speed.Overall, edge computing is not replacing the cloud but complementing it. Together, they form a hybrid architecture that balances performance, scalability, and security. As more devices become intelligent and connected, edge computing will play a crucial role in shaping the next generation of technology-driven systems."
```

```python
import re
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
from nltk.stem import WordNetLemmatizer
```

```python
ps=PorterStemmer
wordnet=WordNetLemmatizer()
sentences=nltk.sent_tokenize(paragraph)
```

```python
sentences
```

```python
len(sentences)
```

```python
corpus=[]
```

```python
for i in range(len(sentences)):
    review=re.sub('[^a-zA-Z]',' ',sentences[i])
    review=review.lower()
    review=review.split()
    review=[wordnet.lemmatize(word) for word in review if not word in set(stopwords.words('english'))]
    review=' '.join(review)
    corpus.append(review)
```

```python
corpus
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer()
X_bow=cv.fit_transform(corpus).toarray()
```

```python
X_bow
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tf=TfidfVectorizer()
X_tf=tf.fit_transform(corpus).toarray()
```
