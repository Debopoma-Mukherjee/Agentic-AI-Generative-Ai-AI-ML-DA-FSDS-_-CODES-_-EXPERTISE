```python
corpus="""Good morning everyone,

Today I want to talk about the importance of continuous learning.

We live in a world where technology changes rapidly. New tools, programming languages, and frameworks appear every year. What we know today may become outdated tomorrow. That is why learning should never stop.

As students and professionals, we often focus on completing exams, assignments, or projects. But true growth happens when we remain curious and keep improving our skills beyond what is required. Reading books, building projects, solving problems, and asking questions all help us become better every day.

In the field of technology, continuous learning is especially important. A person who learns consistently becomes more adaptable, confident, and capable of solving real-world problems. Whether it is machine learning, data engineering, cloud computing, or software development, success comes to those who are willing to learn and practice regularly.

Learning also teaches patience. At first, many concepts seem difficult and confusing. But with persistence and dedication, understanding gradually develops. Every expert was once a beginner who chose not to give up.

So let us make a commitment to learn something new each day. Even small progress, repeated over time, leads to remarkable results.

Thank you.
"""
```

```python
from nltk.stem import PorterStemmer
```

```python
from nltk.corpus import stopwords
```

```python
import nltk
nltk.download('stopwords')
```

```python
stopwords.words('english')
```

```python
from nltk.stem import PorterStemmer
```

```python
stemmer=PorterStemmer()
```

```python
sentence=nltk.sent_tokenize(corpus)
```

```python
for i in range(len(sentence)):
    words=nltk.word_tokenize(sentence[i])
    words=[stemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    sentence[i]=' '.join(words)
```
