```python
import nltk
from nltk.util import bigrams,trigrams,ngrams
```

```python
string ='we are learner of prakash senapati from 9am batch'
quotes_tokens=nltk.word_tokenize(string)
quotes_tokens
```

```python
len(quotes_tokens)
```

```python
quotes_bigrams=list(nltk.bigrams(quotes_tokens))
quotes_bigrams
```

```python
quotes_trigrams=list(nltk.trigrams(quotes_tokens))
quotes_trigrams
```

```python
quotes_ngrams=list(nltk.ngrams(quotes_tokens))
quotes_ngrams
```

```python
quotes_ngrams=list(nltk.ngrams(quotes_tokens,4))
quotes_ngrams
```

```python
quotes_ngrams=list(nltk.ngrams(quotes_tokens,10)) # cant pass 10 as the sentence has only 9 tokens
quotes_ngrams
```
