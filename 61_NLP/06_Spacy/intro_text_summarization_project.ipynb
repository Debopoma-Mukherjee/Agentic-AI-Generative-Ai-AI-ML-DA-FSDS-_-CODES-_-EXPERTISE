# TEXT SUMMARIZATION

```python
text="""
Technology has become an inseparable part of modern life, influencing almost every 
activity we perform each day. From waking up to smartphone alarms to using digital 
platforms for learning, shopping, and communication, technology shapes our routines 
in subtle yet powerful ways. Advances in fields like artificial intelligence, data science,
and cloud computing are transforming industries by improving efficiency and decision-making.
At the same time, these innovations demand continuous learning and adaptability from individuals. 
While technology offers convenience and speed, its true impact depends on how responsibly and 
creatively it is used to solve real-world problem
s and enhance human life."""
```

```python
text
```

```python
import spacy
from spacy.lang.en.stop_words import STOP_WORDS
from string import punctuation
```

```python
stopwords=list(STOP_WORDS)
stopwords
```

```python
len(stopwords)
```

```python
doc=nlp(text)
```

```python
doc
```

```python
tokens=[token.text for token in doc]
print(tokens)
```

```python
word_frequencies={}

for word in doc:
    if word.text.lower() not in stopwords:
        if word.text.lower() not in punctuation:
            if word.text not in word_frequencies.keys():
                word_frequencies[word.text]=1
            else:
                word_frequencies[word.text]+=1
```

```python
word_frequencies
```

```python
len(word_frequencies)
```

```python
max_frequency=max(word_frequencies.values())
```

```python
max_frequency
```

```python
for word in word_frequencies.keys():
    word_frequencies[word]=word_frequencies[word]/max_frequency
```

```python
word_frequencies
```

```python
sentence_tokens =[sent for sent in doc.sents]
sentence_tokens
```

```python
len(sentence_tokens)
```

```python
sentence_scores={}
for sent in sentence_tokens:
    for word in sent:
        if word.text.lower() in word_frequencies.keys():
            if sent not in sentence_scores.keys():
                sentence_scores[sent]=word_frequencies[word.text.lower()]
            else:
                sentence_scores[sent]+=word_frequencies[word.text.lower()]
```

```python
sentence_scores
```

```python
len(sentence_scores)
```

```python
from heapq import nlargest
select_length=int(len(sentence_tokens)*0.4)
select_length
```

```python
summary=nlargest(select_length,sentence_scores,key=sentence_scores.get)
summary
```
