```python
import tensorflow as tf 
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.preprocessing.image import load_img,img_to_array
```

```python
datagen=ImageDataGenerator(
    rotation_range=40,
    width_shift_range=0.4,
    height_shift_range=0.4,
    shear_range=0.4,
    zoom_range=0.4,
    horizontal_flip=True,
    fill_mode='wrap',
    rescale=None)
```

```python
img=load_img(r'C:\Users\91995\NIT\65_Data Augmentation\dog.webp')
img
```

```python
x=img_to_array(img)
x=x.reshape((1,)+x.shape)

i=0
for batch in datagen.flow(x,batch_size=1,save_to_dir=r'C:\Users\91995\NIT\65_Data Augmentation\DATA AUGMENT',save_prefix='dog',save_format='jpeg'):
    i+=1
    if i>30:
        break
```
