# filter


```python
def is_odd(n):
    return n%2!=0
nums=[3,2,6,8,4,6,2,9]
odd=list(filter(is_odd,nums))
print(odd)
```

```python
nums=[3,2,6,8,4,6,2,9]
evens=list(filter(lambda n:n%2==0,nums))
print(evens)
```

```python
nums=[3,2,6,8,4,6,2,9]
evens=list(filter(lambda x:x%2==0,nums))
odds=list(filter(lambda x:x%2!=0,nums))
print(evens)
print(odds)
```

 # map

```python
def update(n):
    return n-2
nums=[3,2,6,8,4,6,2,9]
evens=list(map(update,nums))
print(evens)
```

```python
nums=[3,2,6,8,4,6,2,9]
evens=list(map(lambda n:n+2,nums))
print(evens)
```

# reduce

```python
from functools import reduce
def add_all(a,b):
    return a+b

nums=[3,2,6,8,4,6,2]
evens=list(filter(lambda n:n%2==0,nums))
double=list(map(lambda n:n*2,nums))
sums=reduce(add_all,double)
print(sums)
```

```python
nums=[3,2,6,8,4,6,2]
evens=list(filter(lambda n:n%2==0,nums))
double=list(map(lambda n:n*2,nums))
sums=reduce(lambda a,b:a+b,double)
print(sums)
```
