# variable length argument

```python

def sum(a,b):
    c=a+b
    print(c)
sum(5,6)
```

```python
def sum(a,*b):
    print(type(a))
    print(type(b))
sum(5,6,7,8)
```

```python
def sum(a,*b):
    c=a
    for i in b:
        c=c+i
    print(c)
sum(5,6,7,8,9)
```

```python
def sum(a,*b):
    c=0
    for i in b:
        c=c+i
    print(c)
sum(5,6,7,8,9)
```

# args and kwargs

```python
## args
def person(name,*data):
     print(name)
     print(data)
 person('ALEX',36,'southcity',987767,'phd')
```

```python
## kwargs
def person(name,**data):
     print(name)
     print(data)
person('ALEX',age=36,home_place='southcity',mob=987767,edu='phd')
```

# global variable vs local variable

```python
# The variable declared inside function is local.
# The variable declared outside function is global.
a=10
def something():
    b=15
    print('in function',b)
    print('out function',a)
```

```python
a=10
def something():
    b=15
print('in function',b)
print('out function',a)
```

```python
a=10
def something():
    b=15
    print('in function',b)
print('out function',a)
```

```python
# system by default calls global variable
a=10
def something():
    a=15
print('in function',a)
print('out function',a)
```

```python
#to get o/p local variable we must need to call the function
#if we dont call the function system by default consider the global variable
a=10
def something():
    b=15
    print('in function',b)
something()
print('out function',a)
```

```python
a=10 #global
def something():
    a=55 #local
    print('in function',a)
something()
print('out function',a)
```

```python
a=10
def something():
    global a
    b=15
    print('in function',b)
    print('global variable',a)
something()
print('out function',a)
```

```python
x=10
def update_x():
    global x
    x+=5
update_x()
print(x)
```
