```python
#Function Argument is divided into 2 part
#1)Formal argument
#2)Actual argument
   # a)Postional arguments
   # b)keyword argument
   # c) default argument
   # d)variable length argument
   # e) kwargs(keyword+variable length)
```

```python
def add(x,y):
    c=x+y
    print(c)
add(5,6)
# Here x and y are formal arguments and 5,6 are called actual arguments
```

```python
## Formal and actual argument
def person(name,age):
    print(name)
    print(age)
person('nit',23)
# Here nit and 23 are actual and name and age are formal argument
```

```python
def person(name,age):
    print(name)
    print(age)
person(23,'nit')
# we will still get an output
```

```python
 ## keyword argument
def person(name,age):
    print(name)
    print(age)
person(age=23,name='nit')
```

```python
def person(name,age):
    print(name)
    print(age)
person(age1=23,name='nit')
```

```python
## Default argument
def person(name,age):
    print(age)
person('nit')
```

```python
def person(name,age=18):
    print(age)
person('nit')
```

```python
def person(name,age=18):
    print(age)
person('nit',24)
```
