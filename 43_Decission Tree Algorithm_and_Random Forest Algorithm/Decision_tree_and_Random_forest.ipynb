```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df=pd.read_csv(r"C:\Users\91995\NIT\43_Decission Tree Algorithm_and_Random Forest Algorithm\emp_sal.csv")
print(df)
```

```python
X=df.iloc[:,1:2].values
y=df.iloc[:,2].values
```

```python
# Decission Tree
from sklearn.tree import DecisionTreeRegressor
dt_reg=DecisionTreeRegressor()
dt_reg.fit(X,y)
```

```python
dt_pred=dt_reg.predict([[6.5]])
print(dt_pred)
```

```python
# Decission Tree
from sklearn.tree import DecisionTreeRegressor
dt_reg=DecisionTreeRegressor(criterion="absolute_error",max_depth=3)
dt_reg.fit(X,y)
dt_pred=dt_reg.predict([[6.5]])
print(dt_pred)
```

```python
# Decission Tree
from sklearn.tree import DecisionTreeRegressor
dt_reg=DecisionTreeRegressor(criterion="poisson",max_depth=3)
dt_reg.fit(X,y)
dt_pred=dt_reg.predict([[6.5]])
print(dt_pred)
```

```python
from sklearn.tree import DecisionTreeRegressor
dt_reg=DecisionTreeRegressor(criterion="poisson",max_depth=3,splitter="random")
dt_reg.fit(X,y)
dt_pred=dt_reg.predict([[6.5]])
print(dt_pred)
```
