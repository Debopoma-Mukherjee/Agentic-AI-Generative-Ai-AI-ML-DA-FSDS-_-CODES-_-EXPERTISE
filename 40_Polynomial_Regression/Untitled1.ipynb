```python
# Polynomial Regression
## Simple Linear Regression: y=b0+b1x1
## Multiple Linear Regression y=b0+b1x1+b2x2+...+bnxn
## Polynomial Linear Regression y=b0+b1x1+b2x1^2+...+bnx1^n
import pandas as pd
df=pd.read_csv(r"C:\Users\91995\NIT\36_machine_learning\Polynomial_Regression\emp_sal.csv")
df
```

```python
x=df.iloc[:,1:2].values
y=df.iloc[:,2].values
x
y
```

```python
from sklearn.linear_model import LinearRegression
lin_reg=LinearRegression()
lin_reg.fit(x,y)
```

```python
import matplotlib.pyplot as plt

plt.scatter(x,y,color='red')
plt.plot(x,lin_reg.predict(x),color="blue")
plt.title('Linear Regression graph')
plt.xlabel('Position level')
plt.ylabel('salary')
plt.show()
```

```python
df
```

```python
lin_model_pred=lin_reg.predict([[6.5]])
print(lin_model_pred)
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly_reg=PolynomialFeatures()
x_poly=poly_reg.fit_transform(x)   
poly_reg.fit(x_poly,y)
```

```python
lin_reg_2=LinearRegression()
lin_reg_2.fit(x_poly,y)
```

```python
import matplotlib.pyplot as plt

plt.scatter(x,y,color='red')
plt.plot(x,lin_reg_2.predict(poly_reg.fit_transform(x)),color="blue")
plt.title('Polynomial Regression graph')
plt.xlabel('Position level')
plt.ylabel('salary')
plt.show()
```

```python
poly_model_pred=lin_reg_2.predict(poly_reg.fit_transform([[6.5]]))
poly_model_pred
```

```python
# while we build polynomial model 
## 6.5 prediction degree 1 linear model 3l
## 6.5 prediction degree 2 linear model 1.8l
from sklearn.preprocessing import PolynomialFeatures
poly_reg=PolynomialFeatures(degree=3)
x_poly=poly_reg.fit_transform(x)   
poly_reg.fit(x_poly,y)
lin_reg_2=LinearRegression()
lin_reg_2.fit(x_poly,y)
plt.scatter(x,y,color='red')
plt.plot(x,lin_reg_2.predict(poly_reg.fit_transform(x)),color="blue")
plt.title('Polynomial Regression graph')
plt.xlabel('Position level')
plt.ylabel('salary')
plt.show()
poly_model_pred=lin_reg_2.predict(poly_reg.fit_transform([[6.5]]))
poly_model_pred
```
